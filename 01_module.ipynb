```python
def Extract_number(num):
    digit = []
    while num>0:
        last_number = num%10
        digit.append(last_number)
        num = num//10
    digit.reverse()
        
    return digit

print(Extract_number(1234))
```

```python
def Count(number):
    sum = 0
    while number>0:
        number = number//10
        sum = sum+1
    return sum

print(Count(12345))
```

```python
import math
def Amonsong(number:int)->bool:
    sum=0
    given_number= number
    while number>0:
        last_number = math.fmod(number ,10)**3
        sum = sum+last_number
        number = number//10
    return  given_number==sum

print(Amonsong(153))
```

```python
import math

def amostong_number(number:int)->bool:
    sum = 0
    given_number = number
    while number>0:
        last = math.fmod(number,10)**3
        sum = sum+last
        number = number//10
    return given_number ==  int(sum)

print(amostong_number(153))
```

```python
def reverse_number(number:int)->int:
    sum= 0
    while number>0:
        last = math.fmod(number,10)
        sum=sum*10+last
        number = number//10
    return  int(sum)

print(reverse_number(123))
```

```python
def Palindom_number(number:int)->bool:
    sum= 0
    given_number = number
    while number>0:
        last = math.fmod(number,10)
        sum=sum*10+last
        number = number//10
    return   given_number == int(sum)

print(Palindom_number(121))
```

```python
def Sum_all_Divisor(number:int)->int:
    digit = []
    sum = 0
    for i in range (1,number):
        if number%i:
            digit.append(i)
            sum= sum+i
    return sum,digit

print(Sum_all_Divisor(10))

```

```python
def Eucledian_Algorithm(x,y):
    while x>0 and y>0:
        if x>y:
            x=x%y
        else :
            y = y%x
        
    if x==0:
        return y
    else :
        
        return x

print(Eucledian_Algorithm(13,14))
        
```
